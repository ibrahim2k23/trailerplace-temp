# TrailerPlace LangGraph diagram

This notebook **only visualizes** the compiled graph structure for `TrailerAgentLG` in `src/agent_lg.py`. It does not run a chat turn or call Pinecone.

**Prerequisites:** Open this project from `Chatbot/src` (same folder as `app.py`), with `.env` containing at least `OPENAI_API_KEY` and `PINECONE_API_KEY` (required at import time by `agent_lg`). Editable install: `pip install -e .`


````python
from pathlib import Path

from dotenv import load_dotenv

# Prefer cwd, then ./src, if .env lives outside the immediate cwd
for d in (Path.cwd(), Path.cwd() / "src"):
    env_path = d / ".env"
    if env_path.is_file():
        load_dotenv(env_path)
        break
else:
    load_dotenv()
````

````python
from src.agent_lg import TrailerAgentLG

agent = TrailerAgentLG()
compiled = agent._graph
````

````python
from IPython.display import Markdown, display

graph = compiled.get_graph()
mermaid_src = graph.draw_mermaid()

display(Markdown(f"```mermaid\n{mermaid_src}\n```"))
print("--- Copy below into https://mermaid.live if the diagram does not render ---")
print(mermaid_src)
````

````python
from IPython.display import Image, display

try:
    png_bytes = graph.draw_mermaid_png()
    display(Image(png_bytes))
except Exception as e:
    print(
        "PNG export failed (often missing Graphviz on PATH, or network/API draw). "
        "Use the Mermaid cell above or paste the printed Mermaid into https://mermaid.live\n"
        f"Error: {type(e).__name__}: {e}"
    )
````
